# TABLA DE COMPARATIVA DEL BIG DATA 





| Área                         | Ventajas                                                                                                                                     | Desventajas                                                                                                                                               |
|-----------------------------|----------------------------------------------------------------------------------------------------------------------------------------------|------------------------------------------------------------------------------------------------------------------------------------------------------------|
| Análisis de datos           | Permite procesar grandes volúmenes de datos en tiempo real, generando información útil y oportuna para múltiples sectores. (Sedkaoui, 2018) | La complejidad en la gestión y análisis de datos puede generar errores si no se cuenta con herramientas y personal capacitado. (Holmes, 2018)             |
| Toma de decisiones empresariales | Facilita decisiones más informadas y estratégicas mediante la identificación de patrones, tendencias y comportamientos del mercado. (IBM, 2023) | Riesgo de tomar decisiones sesgadas si los datos están incompletos o mal interpretados. (López Murphy & Zarza, 2017)                                      |
| Privacidad                  | Puede contribuir a detectar fraudes y proteger datos sensibles si se usa de forma ética. (Ríos Insua & Gómez-Ullate, 2019)                 | Amenaza la privacidad de los individuos por la recopilación masiva y, a veces, no autorizada de datos personales. (Holmes, 2018)                          |
| Seguridad                   | Mejora los sistemas de ciberseguridad mediante el análisis predictivo de ataques y vulnerabilidades. (Casas Roma et al., 2019)              | La concentración de grandes volúmenes de datos representa un objetivo atractivo para los ciberdelincuentes. (IBM, 2023)                                   |
| Economía                    | Impulsa la innovación y competitividad al optimizar procesos productivos y reducir costos operativos. (López Murphy & Zarza, 2017)         | Puede aumentar la brecha entre grandes empresas (con acceso a Big Data) y pymes que no cuentan con estos recursos. (Ríos Insua & Gómez-Ullate)           |
| Desarrollo tecnológico      | Estimula el desarrollo de nuevas tecnologías como inteligencia artificial, IoT y aprendizaje automático. (Sedkaoui, 2018)                   | Requiere una infraestructura tecnológica costosa y constante actualización de sistemas y conocimientos. (Casas Roma et al., 2019)                         |
